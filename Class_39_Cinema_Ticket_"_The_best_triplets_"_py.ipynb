## 1. Predict and test · 2 minutes

```
def get_price(x):
    if x < 12:
        return 10
    else:
        return 25
```

## 2. Read the results
Compare the expected and actual prices. Look for FAIL.

```
tests = [(5, 10), (11, 10), (12, 20), (30, 20)]

def run_tests(func):
    for age, expected in tests:
        got = func(age)
        result = "PASS" if got == expected else "FAIL"
        print("age", age, "| should be", expected, "| got", got, "|", result)

run_tests(get_price)
```
